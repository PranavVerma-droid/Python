```python
import pickle

def emp_inp():      #(I)
    e_id = input("Please Enter Employee ID: ")
    e_name = input("Please Enter Employee Name: ")
    e_dept = input("Please Enter Department: ")
    e_salary = int(input("Please Enter Employee Salary: "))

    rec = [e_id, e_name, e_dept, e_salary]

    f = open("Employees.dat", "ab")
    pickle.dump(rec, f)
    f.close()

def update_salary():      #(II)
    rec = []
    f1 = open("Employees.dat", "rb")
    while True:
        try:
            data = pickle.load(f1)
            if data[2] == "IT":
                data[3] = 200000
            rec.append(data)
        except EOFError:
            break
    f1.close()

    f2 = open("Employees.dat", "wb")
    for i in rec:
        pickle.dump(i, f2)
    f2.close()
    
```

```python
def find_shortest_word():
    f = open("textfile.txt", "r")
    data = f.readlines()

    shortest_word = ""
    len_short = float('inf')

    for i in data:
        line_dat = i.strip().split(" ")
        for j in line_dat:
            if len(j) < len_short:
                len_short = len(j)
                shortest_word = j
    
    print("Shortest Word: " + str(shortest_word))
    f.close()
    return shortest_word
```

```python
def gen_ran_bw_range(max, min):
    # Generate a Float Random Number b/w 10 and 20 using random lib
    import random
    max_val = max
    min_val = min

    # Formula:
    # val = random.random() * (max_val - min_val) + min_val

    val = random.random() * (max_val - min_val) + min_val
    print(val)

for i in range(20):
    gen_ran_bw_range(20, 10)
```

```python
def pushstack(stack, character):        #(a)
    stack.append(character)

def popstack(stack):                    #(b)
    if not len(stack):
        print("Stack Underflow.")
    else:
        x = stack.pop()
        return x
    
def isempty(stack):                     #(c)
    if len(stack):
        return False
    else:
        return True

def main():
    stck = []

    str1 = input("Please Enter A String: ")
    print("Input String: " + str1)

    for i in str1:
        pushstack(stck, i)
    
    print("Reversed String: ", end="")
    while not isempty(stck):
        print(popstack(stck), end="")

main()
```

```python
stack = []

def push3_5(n):                         #(i)
    if n % 3 == 0 or n % 5 == 0:
        stack.append(n)

def pop3_5():                           #(ii)
    if not len(stack):
        print("Stack Underflow.")
    else:
        return stack.pop()

def main():
    while True:
        usr_inp = input("Enter 1 to Input & Push Number, 2 to pop element, 3 to exit: ")
        if int(usr_inp) == 1:
            num_inp = int(input("Please Enter Number: "))
            push3_5(num_inp)
        elif int(usr_inp) == 2:
            print("Number Popped: " + str(pop3_5()))
        elif int(usr_inp) == 3:
            break
        else:
            print("Invalid Input.")

main()
```

```python
def display_times():
    f = open("Prog.txt", "r")
    data = f.readlines()
    count = 0
    for i in data:
        line_dat = i.strip().split(" ")
        for j in line_dat:
            if j == "Python":
                count = count + 1
    
    print("Python Word Count: " + str(count))
    f.close()
    return count
```

```python
def Addnew(Book):
    book_name = input("Please Enter Book Name to Add: ")
    Book.append(book_name)
    return book_name

def Remove(Book):
    if not len(Book):
        print("Stack Underflow.")
    else:
        return Book.pop()
    
def main():
    book_stack = []
    while True:
        usr_inp = int(input("Enter 1 to Add Book, 2 to Remove Book, 3 to View Stack, 4 to Exit: "))

        if usr_inp == 1:
            dat = Addnew(book_stack)
            print("Book Added: " + dat)
        elif usr_inp == 2:
            dat = Remove(book_stack)
            print("Book Removed: " + str(dat))
        elif usr_inp == 3:
            print(book_stack)
        elif usr_inp == 4:
            break
        else:
            print("Invalid Input.")

main()
```
