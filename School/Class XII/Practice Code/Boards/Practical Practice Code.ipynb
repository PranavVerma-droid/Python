```python
SHOW DATABASES;

CREATE DATABASE practical;
USE practical;
```

```python
CREATE TABLE Students (
    Rollno      INTEGER         PRIMARY KEY,
    Name        VARCHAR(20)    NOT NULL,
    Gender      CHAR(1),
    Marks       INTEGER(11),
    DOB         DATE
);
```

```python
DESCRIBE Students;
```
